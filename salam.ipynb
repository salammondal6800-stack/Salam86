```python
import os
import pandas as pd

# গিটহাব রিপোজিটরির ভেতরের ডাটা ফোল্ডার পাথ
folder_path = 'data'

# ১৬টি অ্যাসেটের ফাইলের তালিকা
files_config = [
    "nifty_5years_master.parquet",
    "banknifty_5years_master.parquet",
    "finnifty_5years_master.parquet",
    "midcpnifty_5years_master.parquet",
    "sensex_5years_master.parquet",
    "reliance_5years_master.parquet",
    "tcs_5years_master.parquet",
    "hdfcbank_5years_master.parquet",
    "tatasteel_5years_master.parquet",
    "infosys_5years_master.parquet",
    "itc_5years_master.parquet",
    "axisbank_5years_master.parquet",
    "icicibank_5years_master.parquet",
    "sbi_5years_master.parquet",
    "bhartiairtel_5years_master.parquet",
    "tataconsumer_5years_master.parquet"
]

print("==================================================")
print("     ১৬টি ফাইল থেকে শেষ দিনের ক্যান্ডেল ডিলিট শুরু...     ")
print("==================================================")

try:
    for idx, file_name in enumerate(files_config, start=1):
        file_path = os.path.join(folder_path, file_name)

        if os.path.exists(file_path):
            df = pd.read_parquet(file_path)
            if not df.empty:
                # ডেট ও টাইম কলাম থেকে টেম্পোরারি ডেটটাইম তৈরি
                df['datetime_temp'] = pd.to_datetime(df['date'] + ' ' + df['time']).dt.tz_localize(None)

                # ফাইলের একেবারে শেষের লেটেস্ট ডেটটি বের করা
                last_timestamp = df['datetime_temp'].max()
                last_date_str = last_timestamp.strftime('%Y-%m-%d')

                # ওই শেষ তারিখের ডেটাগুলো বাদ দেওয়া (ডিলিট করা)
                df_filtered = df[df['date'] != last_date_str]

                # টেম্পোরারি কলাম বাদ দিয়ে ফাইল সেভ করা
                df_filtered = df_filtered[['date', 'time', 'open', 'high', 'low', 'close', 'volume']]
                df_filtered.to_parquet(file_path, compression='snappy')

                print(f"[{idx}/16] ফাইল: {file_name}")
                print(f"   ❌ সফলভাবে ডিলিট করা হয়েছে: {last_date_str} তারিখের সমস্ত ক্যান্ডেল।")
                print(f"   - বর্তমান মোট ক্যান্ডেল সংখ্যা: {len(df_filtered)}")
            else:
                print(f"[{idx}/16] ফাইল: {file_name} - ফাইলটি খালি রয়েছে।")
        else:
            print(f"[{idx}/16] ফাইল: {file_name} - ফাইলটি পাওয়া যায়নি!")

    print("\n==========================================")
    print("     সব ফাইল থেকে শেষ দিনের ডেটা ডিলিট সম্পন্ন!     ")
    print("==========================================")

    # গিটহাবে অটোমেটিক ফাইল পুশ করার জন্য Git কমান্ডসমূহ
    print("\n🔄 পরিবর্তিত ফাইলগুলো গিটহাবে পুশ করা হচ্ছে...")
    os.system("git config --global user.name 'GitHub Actions Bot'")
    os.system("git config --global user.email 'actions@github.com'")
    os.system("git add data/*.parquet")
    os.system('git commit -m "Delete last day candles from parquet files [skip ci]"')
    os.system("git push")
    print("🎉 গিটহাবে সফলভাবে ডেটা পুশ করা হয়েছে!")

except Exception as e:
    print("একটি ত্রুটি ঘটেছে:", e)
```

```python
import os
import pandas as pd

# গিটহাব রিপোজিটরির ভেতরের ডাটা ফোল্ডার পাথ
folder_path = 'data'

files_config = [
    "nifty_5years_master.parquet",
    "banknifty_5years_master.parquet",
    "finnifty_5years_master.parquet",
    "midcpnifty_5years_master.parquet",
    "sensex_5years_master.parquet",
    "reliance_5years_master.parquet",
    "tcs_5years_master.parquet",
    "hdfcbank_5years_master.parquet",
    "tatasteel_5years_master.parquet",
    "infosys_5years_master.parquet",
    "itc_5years_master.parquet",
    "axisbank_5years_master.parquet",
    "icicibank_5years_master.parquet",
    "sbi_5years_master.parquet",
    "bhartiairtel_5years_master.parquet",
    "tataconsumer_5years_master.parquet"
]

print("==================================================")
print("     ১৬টি ফাইলের ফার্স্ট ডেট, লাস্ট ডেট ও স্ট্যাটাস যাচাই...     ")
print("==================================================")

try:
    for idx, file_name in enumerate(files_config, start=1):
        file_path = os.path.join(folder_path, file_name)

        if os.path.exists(file_path):
            df = pd.read_parquet(file_path)
            if not df.empty:
                # ডেট ও টাইম কলাম থেকে টেম্পোরারি ডেটটাইম তৈরি
                df['datetime_temp'] = pd.to_datetime(df['date'] + ' ' + df['time']).dt.tz_localize(None)

                # প্রথম এবং শেষ ক্যান্ডেলের ডেট-টাইম বের করা
                first_timestamp = df['datetime_temp'].min()
                last_timestamp = df['datetime_temp'].max()

                first_date_str = first_timestamp.strftime('%Y-%m-%d %H:%M:%S')
                last_date_str = last_timestamp.strftime('%Y-%m-%d %H:%M:%S')
                total_candles = len(df)

                print(f"[{idx}/16] ফাইল: {file_name}")
                print(f"   📊 মোট ক্যান্ডেল সংখ্যা: {total_candles}")
                print(f"   📅 প্রথম ডেট (First Date): {first_date_str}")
                print(f"   🕒 শেষ ডেট (Last Date): {last_date_str}")
                print(f"   ----------------------------------------------")
            else:
                print(f"[{idx}/16] ফাইল: {file_name} - ⚠️ ফাইলটি খালি রয়েছে!")
        else:
            print(f"[{idx}/16] ফাইল: {file_name} - ❌ ফাইলটি পাওয়া যায়নি!")

    print("\n==========================================")
    print("     সব ফাইলের যাচাইকরণ সম্পন্ন!     ")
    print("==========================================")

except Exception as e:
    print("একটি ত্রুটি ঘটেছে:", e)
```
